```python
from collections import defaultdict

class Graph:
    def __init__(self):
        self.graph = defaultdict(list)

    def addEdge(self, u, v):
        self.graph[u].append(v)

    def BFS(self, s):
        if not self.graph:
            print("Graph is empty.")
            return

        # Determine the maximum vertex ID to correctly size the visited array.
        all_vertices = set(self.graph.keys())
        for v_list in self.graph.values():
            all_vertices.update(v_list)

        if not all_vertices: # If there are no vertices in the graph
            print(f"No vertices in the graph to start BFS from {s}.")
            return

        max_vertex_id = max(all_vertices)

        visited = [False] * (max_vertex_id + 1)
        queue = []

        if s not in all_vertices and s not in self.graph: # Check if starting vertex exists in graph definition
            print(f"Starting vertex {s} not found in the graph.")
            return

        queue.append(s)
        visited[s] = True

        print(f"Following is Breadth First Traversal (starting from vertex {s})")
        while queue:
            s_curr = queue.pop(0)
            print(s_curr, end=" ")

            # Only iterate through neighbors if s_curr exists as a key in the graph
            if s_curr in self.graph:
                for i in self.graph[s_curr]:
                    if i < len(visited) and not visited[i]: # Ensure i is within bounds of visited
                        queue.append(i)
                        visited[i] = True
        print() # Newline for cleaner output

# Example usage
g = Graph()
g.addEdge(0, 1)
g.addEdge(0, 2)
g.addEdge(1, 2)
g.addEdge(2, 0)
g.addEdge(2, 3)
g.addEdge(3, 3)

g.BFS(2)
```
