```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class CircularLinkedList:
    def __init__(self):
        self.last = None

    def addToEmpty(self, data):
        if self.last is not None:
            return self.last
        temp = Node(data)
        self.last = temp
        self.last.next = self.last
        return self.last

    def addBegin(self, data):
        if self.last is None:
            return self.addToEmpty(data)
        temp = Node(data)
        temp.next = self.last.next
        self.last.next = temp
        return self.last

    def addEnd(self, data):
        if self.last is None:
            return self.addToEmpty(data)
        temp = Node(data)
        temp.next = self.last.next
        self.last.next = temp
        self.last = temp
        return self.last

    def addAfter(self, data, item):
        if self.last is None:
            return None
        temp = Node(data)
        p = self.last.next
        while True:
            if p.data == item:
                temp.next = p.next
                p.next = temp
                if p == self.last:
                    self.last = temp
                return self.last
            p = p.next
            if p == self.last.next:
                print(item, "not present in the list")
                break
        return self.last

    def traverse(self):
        if self.last is None:
            print("List is empty")
            return
        temp = self.last.next
        while True:
            print(temp.data, end = " ")
            temp = temp.next
            if temp == self.last.next:
                break
        print()

if __name__ == '__main__':
    llist = CircularLinkedList()

    last = llist.addToEmpty(6)
    last = llist.addBegin(4)
    last = llist.addBegin(2)
    last = llist.addEnd(8)
    last = llist.addEnd(12)
    last = llist.addAfter(10,8)

    llist.traverse()
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class CreateList:
    def __init__(self):
        self.head = None
        self.tail = None

    def add(self, data):
        newNode = Node(data)
        if self.head is None:
            self.head = newNode
            self.tail = newNode
            newNode.next = self.head  # Points to itself for a single node circular list
        else:
            self.tail.next = newNode
            self.tail = newNode
            self.tail.next = self.head # Connects new tail to head

    def deleteEnd(self):
        if self.head is None:
            print("List is empty, cannot delete.")
            return

        if self.head == self.tail: # Only one node
            self.head = None
            self.tail = None
            return

        current = self.head
        # Find the node before the tail
        while current.next != self.tail:
            current = current.next

        # Now current is the node before the tail
        self.tail = current
        self.tail.next = self.head # New tail points to head

    def display(self):
        if self.head is None:
            print("List is empty")
            return

        current = self.head
        print("Nodes of the circular linked list:")
        while True:
            print(current.data, end=" ")
            current = current.next
            if current == self.head:
                break
        print("\n") # Add a newline at the end

# Instantiate CreateList (which implements the circular linked list)
cl = CreateList()
cl.add(1)
cl.add(2)
cl.add(3)
cl.add(4)

print("Original List:")
cl.display()

print("Deleting elements one by one:")
# Loop to delete elements and display after each deletion
while cl.head is not None:
    cl.deleteEnd()
    print("Updated List:")
    cl.display()
```
